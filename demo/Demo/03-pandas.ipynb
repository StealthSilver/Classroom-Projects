# Pandas — the operations you actually use

Pandas is the Python library for tables. A **DataFrame** is the whole table. A **Series** is one column of that table.

Every part below uses the same class list: `data/students.csv`.

| Column | What it holds |
| --- | --- |
| student_id | id such as S01 |
| name | student name |
| city | home city (a few are blank) |
| department | Computer Science, Mathematics, or Physics |
| subject | the paper they sat |
| score | marks out of 100 (blank if they missed the test) |
| attendance | attendance percent |
| fee | fee in rupees (one value is blank) |
| joined | join date, stored as text until Part 6 |
| status | Active, Alumni, or On leave |

There are 30 students. Farhan Ali (S18) is pasted twice, so the file has 31 rows. A few cells are empty on purpose.

Run the cells from the top. Later parts use the names created earlier (`students`, then `clean`).


## Part 1 — Load the table

`read_csv` reads the file and builds a DataFrame. The first row of the file becomes the column names.


```python
import pandas as pd

pd.set_option("display.max_rows", 40)       # show the whole class list when we ask for it
pd.set_option("display.max_columns", None)   # show every column

# This is the table used in every part below.
students = pd.read_csv("data/students.csv")

students.head()  # first 5 rows
```

```python
students.tail(3)  # last 3 rows — Farhan Ali (S18) is in here twice
```

## Part 2 — Look at what you loaded

Before changing anything, check the size, the column types, and where the blanks are.

`info` counts non-empty cells. `describe` summarizes the number columns: count, mean, min, max. The count there is how many values are filled in, so score and fee count lower than attendance.


```python
print("rows, columns:", students.shape)   # 31 rows, 10 columns
print("columns:", list(students.columns))
students.dtypes                            # the type pandas chose for each column
```

```python
students.info()  # Non-Null Count drops where a cell was left empty
```

```python
students.describe()  # number columns only
```

```python
# unique = the distinct values. nunique = how many. value_counts = how often each one appears.
# Physics is 11 here because Farhan Ali is listed twice. After Part 5 each department has 10.
print(students["department"].unique())
print("departments:", students["department"].nunique())
students["department"].value_counts()
```

## Part 3 — Pick columns and rows

One pair of brackets gives a Series (one column). A list inside the brackets gives a DataFrame (a smaller table).

`loc` picks by **label**. `iloc` picks by **position** (0, 1, 2, …). A `loc` slice includes the end label. An `iloc` slice stops before the end position.


```python
print(type(students["name"]))             # Series — one column
print(type(students[["name", "score"]]))  # DataFrame — still a table
students[["name", "department", "score"]].head()
```

```python
# loc uses labels. The slice 0:4 includes row 4.
print("one cell:", students.loc[0, "name"])
students.loc[0:4, ["name", "department", "score"]]
```

```python
# iloc uses positions. 0:5 is rows 0,1,2,3,4. 0:3 is the first three columns.
students.iloc[0:5, 0:3]
```

```python
# Keep rows where the condition is True. A blank score is left out.
students.loc[students["score"] >= 85, ["name", "department", "subject", "score"]]
```

```python
# & means both conditions must be true. Put parentheses around each one.
students.loc[
    (students["department"] == "Physics") & (students["score"] >= 80),
    ["name", "subject", "score"],
]
```

```python
# isin keeps rows whose value is in the list.
students.loc[
    students["city"].isin(["Pune", "Mumbai"]),
    ["name", "city", "department"],
]
```

## Part 4 — Sort, rename, and use an index

These calls give back a new table. `students` stays as it was, unless you write `students = ...`.


```python
# ascending=False puts the highest score first. Blank scores stay at the bottom.
students.sort_values("score", ascending=False).head(5)
```

```python
# First by department A to Z, then by score high to low inside each department.
students.sort_values(["department", "score"], ascending=[True, False])[
    ["name", "department", "score"]
].head(8)
```

```python
# nlargest is the short way to ask for the top rows of one column.
students.nlargest(5, "score")[["name", "department", "subject", "score"]]
```

```python
renamed = students.rename(columns={"attendance": "attendance_pct", "fee": "fee_inr"})
renamed.head(3)
```

```python
# set_index moves a column into the row label, so loc can look a student up by id.
by_id = students.set_index("student_id")
by_id.loc["S01"]
```

```python
by_id.reset_index().head(3)  # student_id is a normal column again
```

```python
students.drop(columns=["status"]).head(3)  # a new table without the status column
```

## Part 5 — Missing values and duplicate rows

Empty cells in the CSV show up as **NaN** (missing).

We fill city and fee, because those blanks are data-entry gaps. We leave score blank: a made-up mark would change the class average. From here on, `clean` is the table the later parts use. It has 30 rows, one per student.


```python
students.isna().sum()  # how many empty cells in each column
```

```python
# keep=False shows both copies of a repeated row.
students.loc[students.duplicated(keep=False), ["student_id", "name", "subject", "score"]]
```

```python
print("all rows:", students.shape[0])
print("drop a row if any cell is empty:", students.dropna().shape[0])
print("drop a row only when score is empty:", students.dropna(subset=["score"]).shape[0])
```

```python
clean = students.drop_duplicates().copy()  # keep the first S18, drop the second copy

clean["city"] = clean["city"].fillna("Unknown")              # blank city -> Unknown
clean["fee"] = clean["fee"].fillna(clean["fee"].median())    # blank fee -> the middle fee

clean.isna().sum()  # score still has 2 blanks
```

## Part 6 — Column types and dates

`joined` loaded as text. `to_datetime` turns it into a real date, and `.dt` pulls out the year and the month.

`score` stays a decimal column because it still has blanks. A number column with any missing value cannot be a plain integer. After the blank fee is filled, `fee` can become an integer.


```python
clean["joined"] = pd.to_datetime(clean["joined"])  # "2023-07-12" -> a real date
clean["join_year"] = clean["joined"].dt.year
clean["join_month"] = clean["joined"].dt.month

clean[["name", "joined", "join_year", "join_month"]].head()
```

```python
clean["fee"] = clean["fee"].astype(int)  # 43000.0 -> 43000
clean[["fee", "score", "joined"]].dtypes
```

## Part 7 — Make new columns

A new column is a calculation stored under a new name.

- `map` swaps each value using a dictionary
- `cut` places a number into a bin (a grade)
- `.str` runs a text method on every row of a column
- `apply` runs your own function on every value — use it when column math is not enough


```python
clean["fee_in_thousands"] = clean["fee"] / 1000

short_name = {
    "Computer Science": "CS",
    "Mathematics": "Math",
    "Physics": "Phy",
}
clean["dept_code"] = clean["department"].map(short_name)  # full name -> short code

clean[["name", "department", "dept_code", "fee", "fee_in_thousands"]].head()
```

```python
# under 60 -> D, from 60 -> C, from 75 -> B, from 85 -> A. A blank score stays blank.
clean["grade"] = pd.cut(
    clean["score"],
    bins=[0, 60, 75, 85, 100],
    right=False,
    labels=["D", "C", "B", "A"],
)
clean[["name", "score", "grade"]].head(8)
```

```python
clean["first_name"] = clean["name"].str.split().str[0]  # the word before the first space
clean[["name", "first_name"]].head()
```

```python
# str.upper / str.contains work on the whole column at once.
print(clean["name"].str.upper().head(3))
clean.loc[clean["name"].str.contains("Rao"), ["name", "city", "department"]]
```

```python
def result_word(score):
    if pd.isna(score):
        return "Missing"   # a blank mark stays blank
    if score >= 60:
        return "Pass"
    return "Fail"

clean["result"] = clean["score"].apply(result_word)

print(clean["result"].value_counts())
clean.loc[clean["result"] != "Pass", ["name", "subject", "score", "grade", "result"]]
```

## Part 8 — Group and summarize

`groupby` splits the table into groups and calculates one result per group. The mean skips blank scores.

`agg` lets you ask for several calculations in one go. The name on the left is the new column name. The pair on the right is `(which column, which calculation)`.


```python
clean.groupby("department")["score"].mean().round(1)
```

```python
summary = clean.groupby("department").agg(
    students=("student_id", "count"),
    avg_score=("score", "mean"),
    avg_attendance=("attendance", "mean"),
    total_fee=("fee", "sum"),
)
summary.round(1).reset_index()  # reset_index brings department back as a column
```

```python
# One average for each department + subject pair.
clean.groupby(["department", "subject"])["score"].mean().round(1)
```

## Part 9 — Turn a summary into a grid

`pivot_table` is a groupby laid out as a grid: row groups down the side, column groups across the top, one number in each cell.

`margins=True` adds an All row and an All column.


```python
year_scores = clean.pivot_table(
    values="score",
    index="department",
    columns="join_year",
    aggfunc="mean",
    margins=True,        # add All
    margins_name="All",
)
year_scores.round(1)
```

## Part 10 — Combine two tables

`merge` sticks columns on by a matching key, the way a lookup works. The class list comes from the CSV. The department head and building are a small second table typed here, because the CSV does not store them.

`concat` stacks tables that already have the same columns, one under the other.


```python
heads = pd.DataFrame({
    "department": ["Computer Science", "Mathematics", "Physics"],
    "head": ["Dr. Mehta", "Dr. Iyer", "Dr. Khan"],
    "building": ["Block A", "Block B", "Block C"],
})

# how="left" keeps every student. on= is the column the two tables share.
with_heads = clean.merge(heads, on="department", how="left")
with_heads[["name", "department", "head", "building"]].head(6)
```

```python
best = clean.nlargest(2, "score")
weakest = clean.nsmallest(2, "score")

# ignore_index=True writes fresh row numbers 0, 1, 2, 3.
pd.concat([best, weakest], ignore_index=True)[["name", "department", "subject", "score"]]
```

## Part 11 — Save a table back to a file

`to_csv` writes a DataFrame to a CSV. `index=False` leaves out the row numbers, so the file looks like the one we loaded.


```python
clean.to_csv("data/students_cleaned.csv", index=False)
print("saved", clean.shape[0], "rows")
clean.head(3)
```

## Which function to reach for

| You want to… | Use |
| --- | --- |
| Load a CSV | `read_csv` |
| See the first or last rows | `head`, `tail` |
| Size of the table | `shape` |
| Column names and types | `columns`, `dtypes`, `info` |
| Mean, min, max of numbers | `describe` |
| Distinct values | `unique`, `nunique`, `value_counts` |
| One column or a few columns | `df["col"]`, `df[["a", "b"]]` |
| Rows and columns by label | `loc` |
| Rows and columns by position | `iloc` |
| Rows that pass a test | `df.loc[df["col"] > 10]` |
| Value is one of several | `isin` |
| Sort | `sort_values` |
| Top or bottom rows | `nlargest`, `nsmallest` |
| Rename columns | `rename` |
| Look up by an id | `set_index`, then `loc` |
| Put the index back | `reset_index` |
| Remove a column | `drop` |
| Count blanks | `isna` |
| Remove blank rows | `dropna` |
| Fill blanks | `fillna` |
| Find or remove repeated rows | `duplicated`, `drop_duplicates` |
| Real dates | `to_datetime`, then `.dt.year` / `.dt.month` |
| Change a column's type | `astype` |
| Add a column | `df["new"] = ...` |
| Swap values with a dictionary | `map` |
| Turn marks into grade bins | `cut` |
| Text in a column | `.str` |
| Your own function on each value | `apply` |
| One result per group | `groupby`, `agg` |
| That summary as a grid | `pivot_table` |
| Attach a lookup table | `merge` |
| Stack two tables | `concat` |
| Save | `to_csv` |
